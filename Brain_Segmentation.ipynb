```python
# ============================================
# CELL 3: CREATE SIMPLE 3D UNET
# ============================================
print("🏗️ Building 3D U-Net model...")

import torch.nn as nn
import torch.nn.functional as F

class Simple3DUNet(nn.Module):
    """Simple 3D U-Net for brain segmentation"""

    def __init__(self, in_channels=1, out_channels=1, features=32):
        super().__init__()

        # Encoder
        self.enc1 = self._block(in_channels, features)
        self.pool1 = nn.MaxPool3d(2)

        self.enc2 = self._block(features, features * 2)
        self.pool2 = nn.MaxPool3d(2)

        # Bottleneck
        self.bottleneck = self._block(features * 2, features * 4)

        # Decoder
        self.up2 = nn.ConvTranspose3d(features * 4, features * 2, kernel_size=2, stride=2)
        self.dec2 = self._block(features * 4, features * 2)

        self.up1 = nn.ConvTranspose3d(features * 2, features, kernel_size=2, stride=2)
        self.dec1 = self._block(features * 2, features)

        # Output
        self.conv_out = nn.Conv3d(features, out_channels, kernel_size=1)

    def _block(self, in_channels, out_channels):
        return nn.Sequential(
            nn.Conv3d(in_channels, out_channels, 3, padding=1),
            nn.ReLU(inplace=True),
            nn.Conv3d(out_channels, out_channels, 3, padding=1),
            nn.ReLU(inplace=True)
        )

    def forward(self, x):
        # Encoder
        enc1 = self.enc1(x)
        enc2 = self.enc2(self.pool1(enc1))

        # Bottleneck
        bottleneck = self.bottleneck(self.pool2(enc2))

        # Decoder with skip connections
        dec2 = self.up2(bottleneck)
        dec2 = torch.cat((dec2, enc2), dim=1)
        dec2 = self.dec2(dec2)

        dec1 = self.up1(dec2)
        dec1 = torch.cat((dec1, enc1), dim=1)
        dec1 = self.dec1(dec1)

        # Output
        return torch.sigmoid(self.conv_out(dec1))

# Create model
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
model = Simple3DUNet().to(device)

print(f"✅ Model created with {sum(p.numel() for p in model.parameters()):,} parameters")
print(f"🚀 Model moved to: {device}")
```

```python
# ============================================
# COMPLETE 3D U-NET TRAINING PIPELINE
# ============================================

import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import Dataset, DataLoader
import pickle
from pathlib import Path
import numpy as np
from tqdm import tqdm
import matplotlib.pyplot as plt

print("🧠 Setting up Brain Mask Dataset (Skull Stripping)")
print("="*60)

class IXIBrainMaskDataset(Dataset):
    """
    Modified IXI Dataset for whole brain segmentation
    Converts multi-class labels to binary brain mask
    """

    def __init__(self, data_dir, patch_size=(80, 96, 112), normalize=True):
        self.files = sorted(list(Path(data_dir).glob("*.pkl")))
        self.patch_size = patch_size
        self.normalize = normalize
        print(f"   Loaded {len(self.files)} subjects from {data_dir}")

    def __len__(self):
        return len(self.files)

    def __getitem__(self, idx):
        with open(self.files[idx], 'rb') as f:
            data = pickle.load(f)

            # Handle different data structures
            if isinstance(data, tuple) or isinstance(data, list):
                image, label = data[0], data[1]
            elif isinstance(data, dict):
                # Try common key names
                image = data.get('image', data.get('data', data.get('volume', None)))
                label = data.get('label', data.get('seg', data.get('mask', None)))

                if image is None or label is None:
                    # Take first two array-like values
                    arrays = [v for v in data.values() if isinstance(v, (np.ndarray, torch.Tensor))]
                    image, label = arrays[0], arrays[1]
            else:
                raise ValueError(f"Unexpected data format: {type(data)}")

        # Convert to numpy if tensor
        if isinstance(image, torch.Tensor):
            image = image.numpy()
        if isinstance(label, torch.Tensor):
            label = label.numpy()

        # Remove channel dimension if present
        if image.ndim == 4:
            image = image.squeeze(0)
        if label.ndim == 4:
            label = label.squeeze(0)

        # Extract center patch
        image_patch, label_patch = self._extract_center_patch(image, label)

        # **KEY CHANGE: Convert multi-class label to binary brain mask**
        # All non-zero labels become 1 (brain), zero stays 0 (non-brain)
        brain_mask = (label_patch > 0).astype(np.float32)

        # Normalize image
        if self.normalize:
            image_patch = (image_patch - image_patch.mean()) / (image_patch.std() + 1e-8)

        # Convert to torch tensors
        image_tensor = torch.from_numpy(image_patch).unsqueeze(0).float()
        mask_tensor = torch.from_numpy(brain_mask).unsqueeze(0).float()

        return image_tensor, mask_tensor

    def _extract_center_patch(self, image, label):
        """Extract center patch from full volume"""
        d, h, w = image.shape
        pd, ph, pw = self.patch_size

        # Calculate starting indices for center crop
        start_d = max(0, (d - pd) // 2)
        start_h = max(0, (h - ph) // 2)
        start_w = max(0, (w - pw) // 2)

        # Handle case where image is smaller than patch
        end_d = min(d, start_d + pd)
        end_h = min(h, start_h + ph)
        end_w = min(w, start_w + pw)

        # Extract patches
        image_patch = image[start_d:end_d, start_h:end_h, start_w:end_w]
        label_patch = label[start_d:end_d, start_h:end_h, start_w:end_w]

        # Pad if necessary
        if image_patch.shape != self.patch_size:
            pad_d = pd - image_patch.shape[0]
            pad_h = ph - image_patch.shape[1]
            pad_w = pw - image_patch.shape[2]

            image_patch = np.pad(image_patch,
                                ((0, pad_d), (0, pad_h), (0, pad_w)),
                                mode='constant', constant_values=0)
            label_patch = np.pad(label_patch,
                                ((0, pad_d), (0, pad_h), (0, pad_w)),
                                mode='constant', constant_values=0)

        return image_patch, label_patch

print("✅ Dataset class ready!")

# ============================================
# 2. CREATE DATALOADERS
# ============================================

print("\n🏗️ Building Binary Brain Segmentation U-Net...")

class BinaryBrainUNet(nn.Module):
    """
    3D U-Net optimized for binary brain segmentation
    Single output channel for brain vs non-brain
    """

    def __init__(self, in_channels=1, features=32):
        super().__init__()

        # Encoder
        self.enc1 = self._block(in_channels, features)
        self.pool1 = nn.MaxPool3d(2)

        self.enc2 = self._block(features, features * 2)
        self.pool2 = nn.MaxPool3d(2)

        self.enc3 = self._block(features * 2, features * 4)
        self.pool3 = nn.MaxPool3d(2)

        # Bottleneck
        self.bottleneck = self._block(features * 4, features * 8)

        # Decoder
        self.up3 = nn.ConvTranspose3d(features * 8, features * 4, kernel_size=2, stride=2)
        self.dec3 = self._block(features * 8, features * 4)

        self.up2 = nn.ConvTranspose3d(features * 4, features * 2, kernel_size=2, stride=2)
        self.dec2 = self._block(features * 4, features * 2)

        self.up1 = nn.ConvTranspose3d(features * 2, features, kernel_size=2, stride=2)
        self.dec1 = self._block(features * 2, features)

        # Output - single channel for binary segmentation
        self.conv_out = nn.Conv3d(features, 1, kernel_size=1)

    def _block(self, in_channels, out_channels):
        return nn.Sequential(
            nn.Conv3d(in_channels, out_channels, 3, padding=1, bias=False),
            nn.BatchNorm3d(out_channels),
            nn.ReLU(inplace=True),
            nn.Conv3d(out_channels, out_channels, 3, padding=1, bias=False),
            nn.BatchNorm3d(out_channels),
            nn.ReLU(inplace=True)
        )

    def forward(self, x):
        # Encoder
        enc1 = self.enc1(x)
        enc2 = self.enc2(self.pool1(enc1))
        enc3 = self.enc3(self.pool2(enc2))

        # Bottleneck
        bottleneck = self.bottleneck(self.pool3(enc3))

        # Decoder with skip connections
        dec3 = self.up3(bottleneck)
        dec3 = torch.cat((dec3, enc3), dim=1)
        dec3 = self.dec3(dec3)

        dec2 = self.up2(dec3)
        dec2 = torch.cat((dec2, enc2), dim=1)
        dec2 = self.dec2(dec2)

        dec1 = self.up1(dec2)
        dec1 = torch.cat((dec1, enc1), dim=1)
        dec1 = self.dec1(dec1)

        # Output with sigmoid
        return torch.sigmoid(self.conv_out(dec1))

# Create model
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
model = BinaryBrainUNet(features=32).to(device)

print(f"✅ Model created")
print(f"   Parameters: {sum(p.numel() for p in model.parameters()):,}")
print(f"   Device: {device}")
# ============================================
# 3. LOSS FUNCTION & OPTIMIZER
# ============================================

print("\n⚙️ Setting up loss functions...")

class DiceLoss(nn.Module):
    """Dice Loss for binary segmentation"""
    def __init__(self, smooth=1.0):
        super().__init__()
        self.smooth = smooth

    def forward(self, pred, target):
        pred = pred.contiguous().view(-1)
        target = target.contiguous().view(-1)

        intersection = (pred * target).sum()
        dice = (2. * intersection + self.smooth) / (pred.sum() + target.sum() + self.smooth)

        return 1 - dice


class FocalLoss(nn.Module):
    """Focal Loss for handling class imbalance"""
    def __init__(self, alpha=0.25, gamma=2.0):
        super().__init__()
        self.alpha = alpha
        self.gamma = gamma

    def forward(self, pred, target):
        bce = nn.functional.binary_cross_entropy(pred, target, reduction='none')
        pt = torch.exp(-bce)
        focal_loss = self.alpha * (1 - pt) ** self.gamma * bce
        return focal_loss.mean()


class CombinedLoss(nn.Module):
    """Combined Dice + Focal Loss"""
    def __init__(self, dice_weight=0.6, focal_weight=0.4):
        super().__init__()
        self.dice_weight = dice_weight
        self.focal_weight = focal_weight
        self.dice = DiceLoss()
        self.focal = FocalLoss()

    def forward(self, pred, target):
        dice_loss = self.dice(pred, target)
        focal_loss = self.focal(pred, target)
        return self.dice_weight * dice_loss + self.focal_weight * focal_loss

# Setup optimizer and scheduler
criterion = CombinedLoss(dice_weight=0.6, focal_weight=0.4)
optimizer = optim.AdamW(model.parameters(), lr=1e-4, weight_decay=1e-5)
scheduler = optim.lr_scheduler.ReduceLROnPlateau(
    optimizer, mode='max', patience=5, factor=0.5
)

print("✅ Training components ready")
print("   Loss: Dice (60%) + Focal (40%)")
print("   Optimizer: AdamW")
print("   Scheduler: ReduceLROnPlateau")

# ============================================
# 4. TRAINING FUNCTIONS
# ============================================

def dice_coefficient(pred, target, threshold=0.5):
    """Calculate Dice coefficient"""
    pred_binary = (pred > threshold).float()
    smooth = 1.0

    intersection = (pred_binary * target).sum()
    dice = (2. * intersection + smooth) / (pred_binary.sum() + target.sum() + smooth)

    return dice.item()


def train_epoch(model, loader, criterion, optimizer, device):
    """Train for one epoch"""
    model.train()
    total_loss = 0
    total_dice = 0

    pbar = tqdm(loader, desc='Training')
    for images, masks in pbar:
        images = images.to(device)
        masks = masks.to(device)

        # Forward pass
        optimizer.zero_grad()
        outputs = model(images)
        loss = criterion(outputs, masks)

        # Calculate Dice score
        with torch.no_grad():
            dice = dice_coefficient(outputs, masks)
            total_dice += dice

        # Backward pass
        loss.backward()
        optimizer.step()

        total_loss += loss.item()
        pbar.set_postfix({'loss': f'{loss.item():.4f}', 'dice': f'{dice:.4f}'})

    return total_loss / len(loader), total_dice / len(loader)


def validate(model, loader, criterion, device):
    """Validate the model"""
    model.eval()
    total_loss = 0
    total_dice = 0

    with torch.no_grad():
        pbar = tqdm(loader, desc='Validation')
        for images, masks in pbar:
            images = images.to(device)
            masks = masks.to(device)

            outputs = model(images)
            loss = criterion(outputs, masks)
            dice = dice_coefficient(outputs, masks)

            total_loss += loss.item()
            total_dice += dice
            pbar.set_postfix({'loss': f'{loss.item():.4f}', 'dice': f'{dice:.4f}'})

    return total_loss / len(loader), total_dice / len(loader)


def train_brain_mask_model(model, train_loader, val_loader, criterion, optimizer,
                           scheduler, device, num_epochs=20, save_path='best_brain_mask_model.pth'):
    """Complete training loop"""

    print(f"\n🎯 Training Brain Mask Segmentation")
    print(f"   Epochs: {num_epochs}")
    print(f"   Device: {device}")
    print("="*60)

    best_val_dice = 0.0
    history = {
        'train_loss': [], 'val_loss': [],
        'train_dice': [], 'val_dice': []
    }

    for epoch in range(num_epochs):
        print(f"\nEpoch {epoch+1}/{num_epochs}")
        print("-" * 60)

        # Train
        train_loss, train_dice = train_epoch(model, train_loader, criterion, optimizer, device)
        history['train_loss'].append(train_loss)
        history['train_dice'].append(train_dice)

        # Validate
        val_loss, val_dice = validate(model, val_loader, criterion, device)
        history['val_loss'].append(val_loss)
        history['val_dice'].append(val_dice)

        # Scheduler step
        scheduler.step(val_dice)

        # Print summary
        print(f"\nEpoch {epoch+1} Summary:")
        print(f"   Train Loss: {train_loss:.4f} | Train Dice: {train_dice:.4f}")
        print(f"   Val Loss:   {val_loss:.4f} | Val Dice:   {val_dice:.4f}")
        print(f"   LR: {optimizer.param_groups[0]['lr']:.2e}")

        # Save best model
        if val_dice > best_val_dice:
            best_val_dice = val_dice
            torch.save({
                'epoch': epoch,
                'model_state_dict': model.state_dict(),
                'optimizer_state_dict': optimizer.state_dict(),
                'val_dice': val_dice,
                'val_loss': val_loss
            }, save_path)
            print(f"   ✅ New best model! Dice: {val_dice:.4f}")

    print("\n" + "="*60)
    print("🎉 Training complete!")
    print(f"   Best validation Dice: {best_val_dice:.4f}")

    # Plot training curves
    fig, axes = plt.subplots(1, 2, figsize=(15, 5))

    axes[0].plot(history['train_loss'], label='Train Loss')
    axes[0].plot(history['val_loss'], label='Val Loss')
    axes[0].set_xlabel('Epoch')
    axes[0].set_ylabel('Loss')
    axes[0].set_title('Training & Validation Loss')
    axes[0].legend()
    axes[0].grid(True)

    axes[1].plot(history['train_dice'], label='Train Dice')
    axes[1].plot(history['val_dice'], label='Val Dice')
    axes[1].set_xlabel('Epoch')
    axes[1].set_ylabel('Dice Score')
    axes[1].set_title('Training & Validation Dice Score')
    axes[1].legend()
    axes[1].grid(True)

    plt.tight_layout()
    plt.show()

    return history

print("✅ Training functions ready!")

# ============================================
# 5. VISUALIZATION FUNCTION
# ============================================

def visualize_brain_mask(model, dataset, device, idx=0):
    """Visualize brain mask prediction with multiple views"""
    model.eval()

    image, mask = dataset[idx]
    image_input = image.unsqueeze(0).to(device)

    with torch.no_grad():
        pred = model(image_input)

    # Move to CPU and convert to numpy
    image = image.cpu().squeeze(0).numpy()
    mask = mask.cpu().squeeze(0).numpy()
    pred = pred.cpu().squeeze(0).squeeze(0).numpy()

    # Get middle slices
    slice_d = image.shape[0] // 2
    slice_h = image.shape[1] // 2
    slice_w = image.shape[2] // 2

    fig, axes = plt.subplots(3, 4, figsize=(16, 12))

    # Axial view
    axes[0, 0].imshow(image[slice_d], cmap='gray')
    axes[0, 0].set_title('Input (Axial)')
    axes[0, 0].axis('off')

    axes[0, 1].imshow(mask[slice_d], cmap='gray')
    axes[0, 1].set_title('Ground Truth')
    axes[0, 1].axis('off')

    axes[0, 2].imshow(pred[slice_d], cmap='gray')
    axes[0, 2].set_title('Prediction')
    axes[0, 2].axis('off')

    axes[0, 3].imshow(image[slice_d], cmap='gray')
    axes[0, 3].imshow(pred[slice_d] > 0.5, cmap='Reds', alpha=0.5)
    axes[0, 3].set_title('Overlay')
    axes[0, 3].axis('off')

    # Coronal view
    axes[1, 0].imshow(image[:, slice_h, :], cmap='gray')
    axes[1, 0].set_title('Input (Coronal)')
    axes[1, 0].axis('off')

    axes[1, 1].imshow(mask[:, slice_h, :], cmap='gray')
    axes[1, 1].set_title('Ground Truth')
    axes[1, 1].axis('off')

    axes[1, 2].imshow(pred[:, slice_h, :], cmap='gray')
    axes[1, 2].set_title('Prediction')
    axes[1, 2].axis('off')

    axes[1, 3].imshow(image[:, slice_h, :], cmap='gray')
    axes[1, 3].imshow(pred[:, slice_h, :] > 0.5, cmap='Reds', alpha=0.5)
    axes[1, 3].set_title('Overlay')
    axes[1, 3].axis('off')

    # Sagittal view
    axes[2, 0].imshow(image[:, :, slice_w], cmap='gray')
    axes[2, 0].set_title('Input (Sagittal)')
    axes[2, 0].axis('off')

    axes[2, 1].imshow(mask[:, :, slice_w], cmap='gray')
    axes[2, 1].set_title('Ground Truth')
    axes[2, 1].axis('off')

    axes[2, 2].imshow(pred[:, :, slice_w], cmap='gray')
    axes[2, 2].set_title('Prediction')
    axes[2, 2].axis('off')

    axes[2, 3].imshow(image[:, :, slice_w], cmap='gray')
    axes[2, 3].imshow(pred[:, :, slice_w] > 0.5, cmap='Reds', alpha=0.5)
    axes[2, 3].set_title('Overlay')
    axes[2, 3].axis('off')

    # Calculate Dice
    dice = dice_coefficient(torch.from_numpy(pred).unsqueeze(0).unsqueeze(0),
                           torch.from_numpy(mask).unsqueeze(0).unsqueeze(0))

    plt.suptitle(f'Brain Mask Prediction - Dice Score: {dice:.4f}',
                 fontsize=16, fontweight='bold')
    plt.tight_layout()
    plt.show()

print("✅ Visualization function ready!")
# ============================================
# 6. TRAINING LOOP
# ============================================

def train_model(model, train_loader, val_loader, criterion, optimizer,
                scheduler, device, num_epochs=10, save_path='best_model.pth'):
    """Complete training loop"""

    print(f"\n🎯 Starting training for {num_epochs} epochs...")
    print("="*60)

    best_val_loss = float('inf')
    history = {'train_loss': [], 'val_loss': []}

    for epoch in range(num_epochs):
        print(f"\nEpoch {epoch+1}/{num_epochs}")
        print("-" * 60)

        # Train
        train_loss = train_epoch(model, train_loader, criterion, optimizer, device)
        history['train_loss'].append(train_loss)

        # Validate
        val_loss = validate(model, val_loader, criterion, device)
        history['val_loss'].append(val_loss)

        # Scheduler step
        scheduler.step(val_loss)

        # Print summary
        print(f"\nEpoch {epoch+1} Summary:")
        print(f"   Train Loss: {train_loss:.4f}")
        print(f"   Val Loss:   {val_loss:.4f}")
        print(f"   LR: {optimizer.param_groups[0]['lr']:.2e}")

        # Save best model
        if val_loss < best_val_loss:
            best_val_loss = val_loss
            torch.save(model.state_dict(), save_path)
            print(f"   ✅ New best model saved! (val_loss: {val_loss:.4f})")

    print("\n" + "="*60)
    print("🎉 Training complete!")
    print(f"   Best validation loss: {best_val_loss:.4f}")

    # Plot training curves
    plt.figure(figsize=(10, 5))
    plt.plot(history['train_loss'], label='Train Loss')
    plt.plot(history['val_loss'], label='Val Loss')
    plt.xlabel('Epoch')
    plt.ylabel('Loss')
    plt.title('Training History')
    plt.legend()
    plt.grid(True)
    plt.show()

    return history

# ============================================
# 7. READY TO TRAIN
# ============================================

print("\n" + "="*60)
print("🚀 READY TO TRAIN!")
print("="*60)
print("\n📝 Run this to start training:")
print("""
history = train_brain_mask_model(
    model,
    train_loader,
    val_loader,
    criterion,
    optimizer,
    scheduler,
    device,
    num_epochs=20
)
""")

print("\n📝 After training, visualize with:")
print("""
# Load best model
checkpoint = torch.load('best_brain_mask_model.pth')
model.load_state_dict(checkpoint['model_state_dict'])

# Visualize predictions
visualize_brain_mask(model, val_dataset, device, idx=0)
visualize_brain_mask(model, val_dataset, device, idx=1)
visualize_brain_mask(model, val_dataset, device, idx=2)
""")
print("="*60)
```

```python
# ============================================
# COMPLETE SETUP - ALL IN ONE CELL
# ============================================

import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import Dataset, DataLoader
import pickle
from pathlib import Path
import numpy as np
from tqdm import tqdm
import matplotlib.pyplot as plt
import glob
import os

print("🚀 Complete setup starting...")

# ============================================
# 1. DATASET CLASS
# ============================================

class IXIBrainMaskDataset(Dataset):
    """Modified IXI Dataset for whole brain segmentation"""

    def __init__(self, data_dir, patch_size=(80, 96, 112), normalize=True):
        self.files = sorted(list(Path(data_dir).glob("*.pkl")))
        self.patch_size = patch_size
        self.normalize = normalize

    def __len__(self):
        return len(self.files)

    def __getitem__(self, idx):
        with open(self.files[idx], 'rb') as f:
            data = pickle.load(f)

            if isinstance(data, (tuple, list)):
                image, label = data[0], data[1]
            else:
                raise ValueError(f"Unexpected data format: {type(data)}")

        if isinstance(image, torch.Tensor):
            image = image.numpy()
        if isinstance(label, torch.Tensor):
            label = label.numpy()

        if image.ndim == 4:
            image = image.squeeze(0)
        if label.ndim == 4:
            label = label.squeeze(0)

        image_patch, label_patch = self._extract_center_patch(image, label)
        brain_mask = (label_patch > 0).astype(np.float32)

        if self.normalize:
            image_patch = (image_patch - image_patch.mean()) / (image_patch.std() + 1e-8)

        image_tensor = torch.from_numpy(image_patch).unsqueeze(0).float()
        mask_tensor = torch.from_numpy(brain_mask).unsqueeze(0).float()

        return image_tensor, mask_tensor

    def _extract_center_patch(self, image, label):
        d, h, w = image.shape
        pd, ph, pw = self.patch_size

        start_d = max(0, (d - pd) // 2)
        start_h = max(0, (h - ph) // 2)
        start_w = max(0, (w - pw) // 2)

        end_d = min(d, start_d + pd)
        end_h = min(h, start_h + ph)
        end_w = min(w, start_w + pw)

        image_patch = image[start_d:end_d, start_h:end_h, start_w:end_w]
        label_patch = label[start_d:end_d, start_h:end_h, start_w:end_w]

        if image_patch.shape != self.patch_size:
            pad_d = pd - image_patch.shape[0]
            pad_h = ph - image_patch.shape[1]
            pad_w = pw - image_patch.shape[2]

            image_patch = np.pad(image_patch, ((0, pad_d), (0, pad_h), (0, pad_w)), mode='constant')
            label_patch = np.pad(label_patch, ((0, pad_d), (0, pad_h), (0, pad_w)), mode='constant')

        return image_patch, label_patch

print("✅ Dataset class defined")

# ============================================
# 2. MODEL ARCHITECTURE
# ============================================

class BinaryBrainUNet(nn.Module):
    def __init__(self, in_channels=1, features=32):
        super().__init__()
        self.enc1 = self._block(in_channels, features)
        self.pool1 = nn.MaxPool3d(2)
        self.enc2 = self._block(features, features * 2)
        self.pool2 = nn.MaxPool3d(2)
        self.enc3 = self._block(features * 2, features * 4)
        self.pool3 = nn.MaxPool3d(2)
        self.bottleneck = self._block(features * 4, features * 8)
        self.up3 = nn.ConvTranspose3d(features * 8, features * 4, kernel_size=2, stride=2)
        self.dec3 = self._block(features * 8, features * 4)
        self.up2 = nn.ConvTranspose3d(features * 4, features * 2, kernel_size=2, stride=2)
        self.dec2 = self._block(features * 4, features * 2)
        self.up1 = nn.ConvTranspose3d(features * 2, features, kernel_size=2, stride=2)
        self.dec1 = self._block(features * 2, features)
        self.conv_out = nn.Conv3d(features, 1, kernel_size=1)

    def _block(self, in_channels, out_channels):
        return nn.Sequential(
            nn.Conv3d(in_channels, out_channels, 3, padding=1, bias=False),
            nn.BatchNorm3d(out_channels),
            nn.ReLU(inplace=True),
            nn.Conv3d(out_channels, out_channels, 3, padding=1, bias=False),
            nn.BatchNorm3d(out_channels),
            nn.ReLU(inplace=True)
        )

    def forward(self, x):
        enc1 = self.enc1(x)
        enc2 = self.enc2(self.pool1(enc1))
        enc3 = self.enc3(self.pool2(enc2))
        bottleneck = self.bottleneck(self.pool3(enc3))
        dec3 = self.up3(bottleneck)
        dec3 = torch.cat((dec3, enc3), dim=1)
        dec3 = self.dec3(dec3)
        dec2 = self.up2(dec3)
        dec2 = torch.cat((dec2, enc2), dim=1)
        dec2 = self.dec2(dec2)
        dec1 = self.up1(dec2)
        dec1 = torch.cat((dec1, enc1), dim=1)
        dec1 = self.dec1(dec1)
        return torch.sigmoid(self.conv_out(dec1))

print("✅ Model class defined")

# ============================================
# 3. SETUP
# ============================================

device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print(f"✅ Device: {device}")

# Create datasets
train_dataset = IXIBrainMaskDataset('/content/IXI_data/Train')
val_dataset = IXIBrainMaskDataset('/content/IXI_data/Val')
print(f"✅ Datasets: Train={len(train_dataset)}, Val={len(val_dataset)}")

# Create dataloaders
train_loader = DataLoader(train_dataset, batch_size=2, shuffle=True, num_workers=2)
val_loader = DataLoader(val_dataset, batch_size=2, shuffle=False, num_workers=2)
print(f"✅ Dataloaders ready")

# Find model
model_files = glob.glob('/content/*.pth')
if model_files:
    model = BinaryBrainUNet(features=32).to(device)
    checkpoint = torch.load(model_files[0], map_location=device)
    if 'model_state_dict' in checkpoint:
        model.load_state_dict(checkpoint['model_state_dict'])
    else:
        model.load_state_dict(checkpoint)
    model.eval()
    print(f"✅ Model loaded from {os.path.basename(model_files[0])}")
else:
    print("⚠️ No model found - you'll need to train first")
    model = None

print("\n" + "="*60)
print("🎉 READY FOR POST-PROCESSING!")
print("="*60)
```

```python
# ============================================
# SAVE MODEL & TEST ON VALIDATION DATA
# Modified for Brain Skull Stripping
# ============================================

import torch
import matplotlib.pyplot as plt
import numpy as np
import os

print("💾 Saving complete model...")

# Save full model checkpoint
model_path = "/content/brain_skull_stripping_model.pth"
torch.save({
    'model_state_dict': model.state_dict(),
    'optimizer_state_dict': optimizer.state_dict(),
    'model_architecture': 'BinaryBrainUNet',  # CHANGED: Updated architecture name
    'training_info': {
        'final_train_loss': history['train_loss'][-1] if history else None,
        'final_val_loss': history['val_loss'][-1] if history else None,
        'final_train_dice': history['train_dice'][-1] if history else None,  # NEW: Added Dice
        'final_val_dice': history['val_dice'][-1] if history else None,  # NEW: Added Dice
        'best_val_dice': max(history['val_dice']) if history else None,  # CHANGED: Track best Dice, not loss
    }
}, model_path)

print(f"✅ Model saved to: {model_path}")
print(f"   Size: {os.path.getsize(model_path) / (1024**2):.2f} MB")

# ============================================
# COMPREHENSIVE TESTING
# ============================================

print("\n" + "="*60)
print("🔬 COMPREHENSIVE MODEL EVALUATION")
print("="*60)

# Load best model
print("\n📥 Loading best model...")
checkpoint = torch.load('best_brain_mask_model.pth')  # CHANGED: Load from correct file
model.load_state_dict(checkpoint['model_state_dict'])
model.eval()
print(f"✅ Model loaded from epoch {checkpoint['epoch']+1}")
print(f"   Best Dice: {checkpoint['val_dice']:.4f}")

# ============================================
# 1. CALCULATE METRICS ON VALIDATION SET
# ============================================

print("\n📊 Calculating metrics on validation set...")

def calculate_metrics(pred, target, threshold=0.5):
    """Calculate segmentation metrics for binary segmentation"""
    pred_binary = (pred > threshold).float()
    target_binary = target.float()

    # True Positives, False Positives, False Negatives, True Negatives
    tp = (pred_binary * target_binary).sum()
    fp = (pred_binary * (1 - target_binary)).sum()
    fn = ((1 - pred_binary) * target_binary).sum()
    tn = ((1 - pred_binary) * (1 - target_binary)).sum()

    # Dice coefficient
    dice = (2 * tp) / (2 * tp + fp + fn + 1e-8)

    # IoU (Jaccard)
    iou = tp / (tp + fp + fn + 1e-8)

    # Precision and Recall
    precision = tp / (tp + fp + 1e-8)
    recall = tp / (tp + fn + 1e-8)

    # Accuracy
    accuracy = (tp + tn) / (tp + tn + fp + fn + 1e-8)

    # Specificity
    specificity = tn / (tn + fp + 1e-8)

    return {
        'dice': dice.item(),
        'iou': iou.item(),
        'precision': precision.item(),
        'recall': recall.item(),
        'accuracy': accuracy.item(),
        'specificity': specificity.item()  # NEW: Added specificity
    }

# Test on multiple validation samples
num_test_samples = 20  # CHANGED: Test on more samples (20 instead of 10)
all_metrics = []

print(f"   Testing on {num_test_samples} samples...")

for idx in range(min(num_test_samples, len(val_dataset))):
    image, label = val_dataset[idx]
    image = image.unsqueeze(0).to(device)

    with torch.no_grad():
        pred = model(image)

    metrics = calculate_metrics(pred.cpu(), label.unsqueeze(0))
    all_metrics.append(metrics)

# Calculate average metrics
avg_metrics = {
    key: np.mean([m[key] for m in all_metrics])
    for key in all_metrics[0].keys()
}

# Calculate standard deviations
std_metrics = {
    key: np.std([m[key] for m in all_metrics])
    for key in all_metrics[0].keys()
}

print("\n📈 Average Metrics on Validation Set:")
print(f"   Dice Score:  {avg_metrics['dice']:.4f} ± {std_metrics['dice']:.4f}")
print(f"   IoU:         {avg_metrics['iou']:.4f} ± {std_metrics['iou']:.4f}")
print(f"   Precision:   {avg_metrics['precision']:.4f} ± {std_metrics['precision']:.4f}")
print(f"   Recall:      {avg_metrics['recall']:.4f} ± {std_metrics['recall']:.4f}")
print(f"   Specificity: {avg_metrics['specificity']:.4f} ± {std_metrics['specificity']:.4f}")  # NEW
print(f"   Accuracy:    {avg_metrics['accuracy']:.4f} ± {std_metrics['accuracy']:.4f}")

# ============================================
# 2. VISUALIZE MULTIPLE PREDICTIONS
# ============================================

print("\n🖼️ Visualizing predictions on validation samples...")

fig, axes = plt.subplots(3, 4, figsize=(16, 12))  # CHANGED: 4 columns instead of 3

for idx in range(3):
    image, label = val_dataset[idx]
    image_display = image.unsqueeze(0).to(device)

    with torch.no_grad():
        pred = model(image_display)

    # Get middle slice
    slice_idx = image.shape[1] // 2  # Middle of depth dimension

    # Convert to numpy
    img_slice = image[0, slice_idx].cpu().numpy()
    label_slice = label[0, slice_idx].cpu().numpy()
    pred_slice = pred[0, 0, slice_idx].cpu().numpy()

    # Plot - Input
    axes[idx, 0].imshow(img_slice, cmap='gray')
    axes[idx, 0].set_title(f'Sample {idx+1}: Input')
    axes[idx, 0].axis('off')

    # Ground Truth
    axes[idx, 1].imshow(label_slice, cmap='gray')
    axes[idx, 1].set_title('Ground Truth (Brain Mask)')  # CHANGED: More descriptive title
    axes[idx, 1].axis('off')

    # Prediction
    axes[idx, 2].imshow(pred_slice, cmap='gray')
    metrics = calculate_metrics(pred.cpu(), label.unsqueeze(0))
    axes[idx, 2].set_title(f"Prediction (Dice: {metrics['dice']:.3f})")
    axes[idx, 2].axis('off')

    # NEW: Overlay visualization
    axes[idx, 3].imshow(img_slice, cmap='gray')
    axes[idx, 3].imshow(pred_slice > 0.5, cmap='Reds', alpha=0.4)
    axes[idx, 3].set_title('Overlay')
    axes[idx, 3].axis('off')

plt.suptitle('Brain Skull Stripping Results', fontsize=16, fontweight='bold')  # NEW: Added title
plt.tight_layout()
plt.savefig('/content/validation_results_skull_stripping.png', dpi=150, bbox_inches='tight')
print("✅ Saved visualization to: /content/validation_results_skull_stripping.png")
plt.show()

# ============================================
# 3. VISUALIZE 3D VOLUME WITH MULTIPLE VIEWS
# ============================================

print("\n📊 Creating comprehensive 3D volume visualization...")

# Take one sample
image, label = val_dataset[0]
image_display = image.unsqueeze(0).to(device)

with torch.no_grad():
    pred = model(image_display)

# Convert to numpy
img_vol = image[0].cpu().numpy()
label_vol = label[0].cpu().numpy()
pred_vol = pred[0, 0].cpu().numpy()

# CHANGED: Create 4-row visualization (added overlay row)
fig, axes = plt.subplots(4, 5, figsize=(20, 16))

# Show 5 equally spaced slices
depth = img_vol.shape[0]
slice_indices = [int(depth * i / 6) for i in range(1, 6)]

for i, slice_idx in enumerate(slice_indices):
    # Row 1: Input image
    axes[0, i].imshow(img_vol[slice_idx], cmap='gray')
    axes[0, i].set_title(f'Slice {slice_idx}/{depth}')
    axes[0, i].axis('off')

    # Row 2: Ground truth
    axes[1, i].imshow(label_vol[slice_idx], cmap='gray')
    axes[1, i].axis('off')

    # Row 3: Prediction
    axes[2, i].imshow(pred_vol[slice_idx], cmap='gray')
    axes[2, i].axis('off')

    # Row 4: Overlay (NEW)
    axes[3, i].imshow(img_vol[slice_idx], cmap='gray')
    axes[3, i].imshow(pred_vol[slice_idx] > 0.5, cmap='Reds', alpha=0.4)
    axes[3, i].axis('off')

axes[0, 0].set_ylabel('Input MRI', fontsize=14, fontweight='bold')
axes[1, 0].set_ylabel('Ground Truth', fontsize=14, fontweight='bold')
axes[2, 0].set_ylabel('Prediction', fontsize=14, fontweight='bold')
axes[3, 0].set_ylabel('Overlay', fontsize=14, fontweight='bold')  # NEW

plt.suptitle('Brain Skull Stripping - Multi-Slice View', fontsize=16, fontweight='bold')
plt.tight_layout()
plt.savefig('/content/3d_volume_slices_skull_stripping.png', dpi=150, bbox_inches='tight')
print("✅ Saved 3D visualization to: /content/3d_volume_slices_skull_stripping.png")
plt.show()

# ============================================
# 4. MULTI-VIEW ANATOMICAL VISUALIZATION (NEW)
# ============================================

print("\n📊 Creating multi-view anatomical visualization...")

fig, axes = plt.subplots(3, 3, figsize=(15, 15))

# Get middle slices for different views
slice_d = img_vol.shape[0] // 2  # Axial
slice_h = img_vol.shape[1] // 2  # Coronal
slice_w = img_vol.shape[2] // 2  # Sagittal

views = [
    ('Axial', img_vol[slice_d, :, :], label_vol[slice_d, :, :], pred_vol[slice_d, :, :]),
    ('Coronal', img_vol[:, slice_h, :], label_vol[:, slice_h, :], pred_vol[:, slice_h, :]),
    ('Sagittal', img_vol[:, :, slice_w], label_vol[:, :, slice_w], pred_vol[:, :, slice_w])
]

for row, (view_name, img_slice, label_slice, pred_slice) in enumerate(views):
    # Input
    axes[row, 0].imshow(img_slice, cmap='gray')
    axes[row, 0].set_title(f'{view_name} - Input')
    axes[row, 0].axis('off')

    # Ground truth
    axes[row, 1].imshow(label_slice, cmap='gray')
    axes[row, 1].set_title(f'{view_name} - Ground Truth')
    axes[row, 1].axis('off')

    # Prediction with overlay
    axes[row, 2].imshow(img_slice, cmap='gray')
    axes[row, 2].imshow(pred_slice > 0.5, cmap='Reds', alpha=0.4)
    axes[row, 2].set_title(f'{view_name} - Prediction')
    axes[row, 2].axis('off')

plt.suptitle('Brain Skull Stripping - Anatomical Views', fontsize=16, fontweight='bold')
plt.tight_layout()
plt.savefig('/content/anatomical_views_skull_stripping.png', dpi=150, bbox_inches='tight')
print("✅ Saved anatomical views to: /content/anatomical_views_skull_stripping.png")
plt.show()

# ============================================
# 5. SUMMARY
# ============================================

print("\n" + "="*60)
print("✅ EVALUATION COMPLETE!")
print("="*60)

print("\n📋 Summary:")
print(f"   Model: Binary Brain U-Net (Skull Stripping)")  # CHANGED
print(f"   Architecture: 3-level encoder/decoder with BatchNorm")  # NEW
print(f"   Parameters: {sum(p.numel() for p in model.parameters()):,}")
print(f"   Input size: (1, 80, 96, 112)")
print(f"   Output size: (1, 80, 96, 112)")
print(f"   Task: Whole brain segmentation (brain vs non-brain)")  # NEW

print(f"\n   📊 Performance Metrics:")
print(f"   ├─ Dice Score:   {avg_metrics['dice']:.4f} ± {std_metrics['dice']:.4f}")
print(f"   ├─ IoU:          {avg_metrics['iou']:.4f} ± {std_metrics['iou']:.4f}")
print(f"   ├─ Precision:    {avg_metrics['precision']:.4f} ± {std_metrics['precision']:.4f}")
print(f"   ├─ Recall:       {avg_metrics['recall']:.4f} ± {std_metrics['recall']:.4f}")
print(f"   ├─ Specificity:  {avg_metrics['specificity']:.4f} ± {std_metrics['specificity']:.4f}")
print(f"   └─ Accuracy:     {avg_metrics['accuracy']:.4f} ± {std_metrics['accuracy']:.4f}")

print("\n📁 Saved Files:")
print("   ├─ Model checkpoint: /content/brain_skull_stripping_model.pth")
print("   ├─ Best model weights: /content/best_brain_mask_model.pth")
print("   ├─ Validation results: /content/validation_results_skull_stripping.png")
print("   ├─ 3D volume slices: /content/3d_volume_slices_skull_stripping.png")
print("   └─ Anatomical views: /content/anatomical_views_skull_stripping.png")

# NEW: Performance interpretation
print("\n💡 Performance Interpretation:")
if avg_metrics['dice'] > 0.98:
    print("   🌟 EXCELLENT: Dice > 0.98 - Clinical grade skull stripping!")
elif avg_metrics['dice'] > 0.95:
    print("   ✅ VERY GOOD: Dice > 0.95 - High quality brain extraction")
elif avg_metrics['dice'] > 0.90:
    print("   👍 GOOD: Dice > 0.90 - Acceptable performance")
else:
    print("   ⚠️ FAIR: Consider further training or hyperparameter tuning")

print("\n🎯 What Your Model Does:")
print("   ✓ Removes skull, eyes, and non-brain tissue")
print("   ✓ Preserves all brain tissue (gray + white matter)")
print("   ✓ Binary segmentation: 1 = brain, 0 = non-brain")
print("   ✓ Works on 3D MRI volumes")

print("\n💡 Next Steps:")
print("   1. Test on held-out test set for final evaluation")
print("   2. Apply to full-resolution images (not just patches)")
print("   3. Post-processing: morphological operations for cleanup")
print("   4. Compare with traditional methods (FSL BET, FreeSurfer)")
print("   5. Deploy for preprocessing pipeline in neuroimaging studies")

print("\n🎉 Brain Skull Stripping Project Complete!")
print("="*60)
```

```python
# ============================================
# LOAD TRAINED MODEL
# ============================================

import torch
import torch.nn as nn
from torch.utils.data import DataLoader

print("📥 Loading trained model...")

# Recreate the model architecture
class BinaryBrainUNet(nn.Module):
    """3D U-Net for binary brain segmentation"""
    def __init__(self, in_channels=1, features=32):
        super().__init__()

        # Encoder
        self.enc1 = self._block(in_channels, features)
        self.pool1 = nn.MaxPool3d(2)

        self.enc2 = self._block(features, features * 2)
        self.pool2 = nn.MaxPool3d(2)

        self.enc3 = self._block(features * 2, features * 4)
        self.pool3 = nn.MaxPool3d(2)

        # Bottleneck
        self.bottleneck = self._block(features * 4, features * 8)

        # Decoder
        self.up3 = nn.ConvTranspose3d(features * 8, features * 4, kernel_size=2, stride=2)
        self.dec3 = self._block(features * 8, features * 4)

        self.up2 = nn.ConvTranspose3d(features * 4, features * 2, kernel_size=2, stride=2)
        self.dec2 = self._block(features * 4, features * 2)

        self.up1 = nn.ConvTranspose3d(features * 2, features, kernel_size=2, stride=2)
        self.dec1 = self._block(features * 2, features)

        # Output
        self.conv_out = nn.Conv3d(features, 1, kernel_size=1)

    def _block(self, in_channels, out_channels):
        return nn.Sequential(
            nn.Conv3d(in_channels, out_channels, 3, padding=1, bias=False),
            nn.BatchNorm3d(out_channels),
            nn.ReLU(inplace=True),
            nn.Conv3d(out_channels, out_channels, 3, padding=1, bias=False),
            nn.BatchNorm3d(out_channels),
            nn.ReLU(inplace=True)
        )

    def forward(self, x):
        # Encoder
        enc1 = self.enc1(x)
        enc2 = self.enc2(self.pool1(enc1))
        enc3 = self.enc3(self.pool2(enc2))

        # Bottleneck
        bottleneck = self.bottleneck(self.pool3(enc3))

        # Decoder
        dec3 = self.up3(bottleneck)
        dec3 = torch.cat((dec3, enc3), dim=1)
        dec3 = self.dec3(dec3)

        dec2 = self.up2(dec3)
        dec2 = torch.cat((dec2, enc2), dim=1)
        dec2 = self.dec2(dec2)

        dec1 = self.up1(dec2)
        dec1 = torch.cat((dec1, enc1), dim=1)
        dec1 = self.dec1(dec1)

        return torch.sigmoid(self.conv_out(dec1))

# Setup device
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')

# Create model
model = BinaryBrainUNet(features=32).to(device)

# Load best weights
checkpoint = torch.load('best_brain_mask_model.pth')
model.load_state_dict(checkpoint['model_state_dict'])
model.eval()

print(f"✅ Model loaded successfully!")
print(f"   Epoch: {checkpoint['epoch']+1}")
print(f"   Best Dice: {checkpoint['val_dice']:.4f}")
print(f"   Device: {device}")

# Also need to recreate val_dataset if not already loaded
from pathlib import Path
import pickle
import numpy as np

class IXIBrainMaskDataset:
    """Dataset for binary brain mask"""
    def __init__(self, data_dir, patch_size=(80, 96, 112), normalize=True):
        self.files = sorted(list(Path(data_dir).glob("*.pkl")))
        self.patch_size = patch_size
        self.normalize = normalize

    def __len__(self):
        return len(self.files)

    def __getitem__(self, idx):
        with open(self.files[idx], 'rb') as f:
            data = pickle.load(f)
            if isinstance(data, tuple) or isinstance(data, list):
                image, label = data[0], data[1]
            else:
                raise ValueError(f"Unexpected data format")

        if isinstance(image, torch.Tensor):
            image = image.numpy()
        if isinstance(label, torch.Tensor):
            label = label.numpy()

        if image.ndim == 4:
            image = image.squeeze(0)
        if label.ndim == 4:
            label = label.squeeze(0)

        image_patch, label_patch = self._extract_center_patch(image, label)
        brain_mask = (label_patch > 0).astype(np.float32)

        if self.normalize:
            image_patch = (image_patch - image_patch.mean()) / (image_patch.std() + 1e-8)

        image_tensor = torch.from_numpy(image_patch).unsqueeze(0).float()
        mask_tensor = torch.from_numpy(brain_mask).unsqueeze(0).float()

        return image_tensor, mask_tensor

    def _extract_center_patch(self, image, label):
        d, h, w = image.shape
        pd, ph, pw = self.patch_size

        start_d = max(0, (d - pd) // 2)
        start_h = max(0, (h - ph) // 2)
        start_w = max(0, (w - pw) // 2)

        end_d = min(d, start_d + pd)
        end_h = min(h, start_h + ph)
        end_w = min(w, start_w + pw)

        image_patch = image[start_d:end_d, start_h:end_h, start_w:end_w]
        label_patch = label[start_d:end_d, start_h:end_h, start_w:end_w]

        if image_patch.shape != self.patch_size:
            pad_d = pd - image_patch.shape[0]
            pad_h = ph - image_patch.shape[1]
            pad_w = pw - image_patch.shape[2]

            image_patch = np.pad(image_patch, ((0, pad_d), (0, pad_h), (0, pad_w)), mode='constant')
            label_patch = np.pad(label_patch, ((0, pad_d), (0, pad_h), (0, pad_w)), mode='constant')

        return image_patch, label_patch

# Load validation dataset
val_dataset = IXIBrainMaskDataset('/content/IXI_data/Val')

print(f"\n✅ Validation dataset loaded: {len(val_dataset)} samples")
print("\n" + "="*60)
print("🚀 Ready for post-processing!")
print("="*60)
```

```python
# ============================================
# FULL VOLUME INFERENCE (NO CROPPING)
# ============================================

import torch
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path
import pickle

print("🧠 Setting up full volume inference...")

# ============================================
# SLIDING WINDOW INFERENCE
# ============================================

def predict_full_volume(model, image_volume, device, patch_size=(80, 96, 112),
                        overlap=16, batch_size=4):
    """
    Predict on full volume using sliding window approach

    Args:
        model: Trained model
        image_volume: Full 3D image (D, H, W)
        device: torch device
        patch_size: Size of patches to process
        overlap: Overlap between patches (higher = smoother but slower)
        batch_size: Number of patches to process at once

    Returns:
        Full volume prediction (same size as input)
    """
    model.eval()

    # Normalize image
    image_normalized = (image_volume - image_volume.mean()) / (image_volume.std() + 1e-8)

    D, H, W = image_volume.shape
    pd, ph, pw = patch_size

    # Initialize output and count arrays
    output = np.zeros((D, H, W), dtype=np.float32)
    count = np.zeros((D, H, W), dtype=np.float32)

    # Calculate stride
    stride_d = max(1, pd - overlap)
    stride_h = max(1, ph - overlap)
    stride_w = max(1, pw - overlap)

    # Collect all patch coordinates
    patches = []
    coords = []

    print(f"   Volume size: {D}x{H}x{W}")
    print(f"   Patch size: {pd}x{ph}x{pw}")
    print(f"   Stride: {stride_d}x{stride_h}x{stride_w}")

    # Generate all patch locations
    for d in range(0, D - pd + 1, stride_d):
        for h in range(0, H - ph + 1, stride_h):
            for w in range(0, W - pw + 1, stride_w):
                patch = image_normalized[d:d+pd, h:h+ph, w:w+pw]
                patches.append(patch)
                coords.append((d, h, w))

    # Handle edge cases (patches that don't fit)
    # Add patches at the edges if needed
    if D > pd:
        d_last = D - pd
        for h in range(0, H - ph + 1, stride_h):
            for w in range(0, W - pw + 1, stride_w):
                patch = image_normalized[d_last:d_last+pd, h:h+ph, w:w+pw]
                patches.append(patch)
                coords.append((d_last, h, w))

    if H > ph:
        h_last = H - ph
        for d in range(0, D - pd + 1, stride_d):
            for w in range(0, W - pw + 1, stride_w):
                patch = image_normalized[d:d+pd, h_last:h_last+ph, w:w+pw]
                patches.append(patch)
                coords.append((d, h_last, w))

    if W > pw:
        w_last = W - pw
        for d in range(0, D - pd + 1, stride_d):
            for h in range(0, H - ph + 1, stride_h):
                patch = image_normalized[d:d+pd, h:h+ph, w_last:w_last+pw]
                patches.append(patch)
                coords.append((d, h, w_last))

    total_patches = len(patches)
    print(f"   Total patches: {total_patches}")

    # Process patches in batches
    with torch.no_grad():
        for i in range(0, total_patches, batch_size):
            batch_patches = patches[i:i+batch_size]
            batch_coords = coords[i:i+batch_size]

            # Convert to tensor
            batch_tensor = torch.stack([
                torch.from_numpy(p).unsqueeze(0).float()
                for p in batch_patches
            ]).to(device)

            # Predict
            batch_pred = model(batch_tensor).cpu().numpy()

            # Aggregate predictions
            for pred, (d, h, w) in zip(batch_pred, batch_coords):
                pred = pred.squeeze()  # Remove channel dimension
                output[d:d+pd, h:h+ph, w:w+pw] += pred
                count[d:d+pd, h:h+ph, w:w+pw] += 1

            if (i // batch_size) % 10 == 0:
                print(f"   Processed {i}/{total_patches} patches...", end='\r')

    print(f"   Processed {total_patches}/{total_patches} patches... Done!")

    # Average overlapping predictions
    output = output / (count + 1e-8)

    return output


# ============================================
# VISUALIZE FULL VOLUME PREDICTION
# ============================================

def visualize_full_volume_prediction(image, prediction, ground_truth=None,
                                     threshold=0.5, save_path=None):
    """
    Visualize full volume prediction with multiple slices
    """
    D, H, W = image.shape
    pred_binary = (prediction > threshold).astype(float)

    # Select slices to show (evenly spaced)
    num_slices = 5
    slice_indices = [int(D * i / (num_slices + 1)) for i in range(1, num_slices + 1)]

    if ground_truth is not None:
        # With ground truth
        fig, axes = plt.subplots(num_slices, 4, figsize=(16, 4*num_slices))

        for idx, slice_idx in enumerate(slice_indices):
            # Input
            axes[idx, 0].imshow(image[slice_idx], cmap='gray')
            axes[idx, 0].set_title(f'Input - Slice {slice_idx}/{D}')
            axes[idx, 0].axis('off')

            # Ground truth
            axes[idx, 1].imshow(ground_truth[slice_idx], cmap='gray')
            axes[idx, 1].set_title('Ground Truth')
            axes[idx, 1].axis('off')

            # Prediction
            axes[idx, 2].imshow(pred_binary[slice_idx], cmap='gray')
            axes[idx, 2].set_title('Prediction')
            axes[idx, 2].axis('off')

            # Overlay
            axes[idx, 3].imshow(image[slice_idx], cmap='gray')
            axes[idx, 3].imshow(pred_binary[slice_idx], cmap='Reds', alpha=0.4)
            axes[idx, 3].set_title('Overlay')
            axes[idx, 3].axis('off')
    else:
        # Without ground truth (for tumor data)
        fig, axes = plt.subplots(num_slices, 3, figsize=(12, 4*num_slices))

        for idx, slice_idx in enumerate(slice_indices):
            # Input
            axes[idx, 0].imshow(image[slice_idx], cmap='gray')
            axes[idx, 0].set_title(f'Input - Slice {slice_idx}/{D}')
            axes[idx, 0].axis('off')

            # Prediction
            axes[idx, 1].imshow(pred_binary[slice_idx], cmap='gray')
            axes[idx, 1].set_title('Brain Mask Prediction')
            axes[idx, 1].axis('off')

            # Overlay
            axes[idx, 2].imshow(image[slice_idx], cmap='gray')
            axes[idx, 2].imshow(pred_binary[slice_idx], cmap='Reds', alpha=0.4)
            axes[idx, 2].set_title('Overlay')
            axes[idx, 2].axis('off')

    plt.tight_layout()

    if save_path:
        plt.savefig(save_path, dpi=150, bbox_inches='tight')
        print(f"   Saved to: {save_path}")

    plt.show()

    # Calculate statistics
    brain_volume = pred_binary.sum()
    total_volume = pred_binary.size
    brain_percentage = (brain_volume / total_volume) * 100

    print(f"\n📊 Statistics:")
    print(f"   Volume size: {D}x{H}x{W}")
    print(f"   Brain voxels: {int(brain_volume):,}")
    print(f"   Total voxels: {total_volume:,}")
    print(f"   Brain percentage: {brain_percentage:.2f}%")


# ============================================
# TEST ON IXI FULL VOLUMES
# ============================================

def test_on_ixi_full_volume(model, device, dataset_path='/content/IXI_data/Val',
                            subject_idx=0):
    """
    Test on a full IXI volume without cropping
    """
    print(f"\n{'='*60}")
    print("🧪 Testing on Full IXI Volume")
    print('='*60)

    # Load full volume
    val_files = sorted(list(Path(dataset_path).glob("*.pkl")))

    if subject_idx >= len(val_files):
        print(f"❌ Subject index {subject_idx} out of range (max: {len(val_files)-1})")
        return

    print(f"\n📂 Loading: {val_files[subject_idx].name}")

    with open(val_files[subject_idx], 'rb') as f:
        data = pickle.load(f)
        if isinstance(data, (tuple, list)):
            image, label = data[0], data[1]
        else:
            raise ValueError("Unexpected data format")

    # Convert to numpy if needed
    if isinstance(image, torch.Tensor):
        image = image.numpy()
    if isinstance(label, torch.Tensor):
        label = label.numpy()

    # Remove channel dimension if present
    if image.ndim == 4:
        image = image.squeeze(0)
    if label.ndim == 4:
        label = label.squeeze(0)

    print(f"   Image shape: {image.shape}")
    print(f"   Label shape: {label.shape}")

    # Convert label to binary
    ground_truth = (label > 0).astype(float)

    # Predict
    print("\n🔮 Running inference...")
    prediction = predict_full_volume(model, image, device,
                                     patch_size=(80, 96, 112),
                                     overlap=16, batch_size=4)

    # Calculate Dice
    pred_binary = (prediction > 0.5).astype(float)
    intersection = (pred_binary * ground_truth).sum()
    dice = (2. * intersection) / (pred_binary.sum() + ground_truth.sum() + 1e-8)

    print(f"\n✅ Dice Score: {dice:.4f}")

    # Visualize
    visualize_full_volume_prediction(
        image, prediction, ground_truth,
        save_path=f'/content/full_volume_ixi_{subject_idx}.png'
    )

    return image, prediction, ground_truth


# ============================================
# DOWNLOAD & TEST ON BRAIN TUMOR DATA
# ============================================

def setup_tumor_dataset():
    """
    Download BraTS (Brain Tumor) dataset sample
    """
    print(f"\n{'='*60}")
    print("🧠 Setting up Brain Tumor Dataset")
    print('='*60)

    print("\n📥 Downloading BraTS sample data...")

    # Download a sample from BraTS dataset
    # Note: This is a simplified version. For full BraTS, you'd need kaggle credentials

    print("""
    To test on brain tumor data, you have two options:

    1. Upload your own tumor MRI files:
       - Click the folder icon on the left
       - Upload .nii.gz files to /content/tumor_data/

    2. Use a public dataset:
       Run these commands:
    """)

    print("""
# Option 2: Download BraTS sample
!pip install -q kaggle
!mkdir -p ~/.kaggle
# Upload your kaggle.json to /content/
!cp /content/kaggle.json ~/.kaggle/
!chmod 600 ~/.kaggle/kaggle.json

# Download BraTS dataset
!kaggle datasets download -d awsaf49/brats2020-training-data
!unzip -q brats2020-training-data.zip -d /content/brats_data/
    """)


def test_on_tumor_volume(model, device, tumor_path):
    """
    Test on brain tumor MRI
    """
    print(f"\n{'='*60}")
    print("🔬 Testing on Brain Tumor Volume")
    print('='*60)

    import nibabel as nib

    # Load tumor image
    print(f"\n📂 Loading: {tumor_path}")

    try:
        img_nii = nib.load(tumor_path)
        image = img_nii.get_fdata()

        print(f"   Image shape: {image.shape}")
        print(f"   Image range: [{image.min():.2f}, {image.max():.2f}]")

        # Handle 4D images (take first volume)
        if image.ndim == 4:
            print(f"   4D image detected, taking first volume")
            image = image[..., 0]

        # Predict
        print("\n🔮 Running inference on tumor brain...")
        prediction = predict_full_volume(model, image, device,
                                         patch_size=(80, 96, 112),
                                         overlap=16, batch_size=4)

        # Visualize
        visualize_full_volume_prediction(
            image, prediction, ground_truth=None,
            save_path='/content/tumor_brain_prediction.png'
        )

        print("\n✅ Tumor brain segmentation complete!")
        print("   Check if the model correctly identifies:")
        print("   - Brain tissue (including tumor)")
        print("   - vs. skull, eyes, and non-brain structures")

        return image, prediction

    except Exception as e:
        print(f"❌ Error: {e}")
        return None, None


# ============================================
# READY TO USE
# ============================================

print("✅ Full volume inference ready!")
print("\n📝 Usage:")
print("""
# 1. Test on full IXI volume (no cropping):
image, pred, gt = test_on_ixi_full_volume(model, device, subject_idx=0)

# 2. Try multiple IXI subjects:
for i in range(3):
    test_on_ixi_full_volume(model, device, subject_idx=i)

# 3. Setup tumor dataset:
setup_tumor_dataset()

# 4. Test on tumor brain (after uploading tumor MRI):
tumor_path = '/content/tumor_data/your_tumor_scan.nii.gz'
image, pred = test_on_tumor_volume(model, device, tumor_path)
""")
```
